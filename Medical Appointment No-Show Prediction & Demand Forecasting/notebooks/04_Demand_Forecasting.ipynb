# Demand Forecasting

Train and compare LightGBM (lag-features), SARIMAX, and LSTM. Outputs final metrics and saves models to `models/`.

```python
from pathlib import Path
from src.models import demand_forecasting

PROCESSED = Path('..') / 'data' / 'processed_medical_appointments.csv'
if not PROCESSED.exists():
    print('Processed data not found at', PROCESSED)
else:
    df = demand_forecasting.load_processed(str(PROCESSED))
    daily = demand_forecasting.aggregate_daily(df)
    print('Daily series length:', len(daily))
    print('
Training LightGBM...')
    try:
        lgb_metrics = demand_forecasting.train_lightgbm(daily, save_dir=str(Path('..') / 'models'))
        print('LightGBM metrics:', lgb_metrics)
    except Exception as e:
        print('LightGBM error:', e)

    print('
Training SARIMAX...')
    try:
        sarimax_metrics = demand_forecasting.train_sarimax(daily, save_dir=str(Path('..') / 'models'))
        print('SARIMAX metrics:', sarimax_metrics)
    except Exception as e:
        print('SARIMAX error:', e)

    print('
Training LSTM...')
    try:
        lstm_metrics = demand_forecasting.train_lstm(daily, save_dir=str(Path('..') / 'models'))
        print('LSTM metrics:', lstm_metrics)
    except Exception as e:
        print('LSTM error (likely missing TF):', e)
```
